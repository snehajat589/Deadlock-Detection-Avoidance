```python

# Deadlock Detection and Avoidance Simulator
# Project: Operating System

print("=" * 55)
print(" DEADLOCK DETECTION AND AVOIDANCE SIMULATOR ")
print("        Using Banker's Algorithm")
print("=" * 55)

# Number of processes and resources
n = 5
m = 3

# Allocation Matrix
allocation = [
    [0, 1, 0],
    [2, 0, 0],
    [3, 0, 2],
    [2, 1, 1],
    [0, 0, 2]
]

# Maximum Resource Matrix
maximum = [
    [7, 5, 3],
    [3, 2, 2],
    [9, 0, 2],
    [2, 2, 2],
    [4, 3, 3]
]

# Available Resources
available = [3, 3, 2]

print("\nNumber of Processes:", n)
print("Number of Resources:", m)
print("Resources: A, B, C")

print("\nAllocation Matrix:")
for row in allocation:
    print(row)

print("\nMaximum Matrix:")
for row in maximum:
    print(row)

print("\nAvailable Resources:", available)
```

```python

# Step 2: Calculate Need Matrix

need = []

for i in range(n):
    row = []
    for j in range(m):
        value = maximum[i][j] - allocation[i][j]
        row.append(value)
    need.append(row)

print("=" * 40)
print("          NEED MATRIX")
print("=" * 40)

print("Process    A    B    C")

for i in range(n):
    print(f"P{i}         ", end="")
    for j in range(m):
        print(need[i][j], end="    ")
    print()
```

```python

# Step 3: Banker's Algorithm - Safe State Check

work = available.copy()
finish = [False] * n
safe_sequence = []

while len(safe_sequence) < n:
    found = False

    for i in range(n):
        if not finish[i]:
            possible = True

            for j in range(m):
                if need[i][j] > work[j]:
                    possible = False
                    break

            if possible:
                print(f"P{i} can execute. Available before execution: {work}")

                for j in range(m):
                    work[j] += allocation[i][j]

                finish[i] = True
                safe_sequence.append(i)
                found = True

                print(f"P{i} completed. Available after execution: {work}")
                print()

    if not found:
        break

print("=" * 45)

if len(safe_sequence) == n:
    print("System is in a SAFE STATE.")
    print("Safe Sequence:", " -> ".join(f"P{i}" for i in safe_sequence))
else:
    print("System is in an UNSAFE STATE.")
    print("No complete safe sequence exists.")

print("=" * 45)
```

```python

# Step 5: Banker’s Algorithm Integration

print("\n" + "=" * 50)
print("        BANKER'S ALGORITHM RESULT")
print("=" * 50)

# Initialize working resources
work = available.copy()
finish = [False] * n
safe_sequence = []

print("\nInitial Available Resources:", work)

while len(safe_sequence) < n:
    found = False

    for i in range(n):
        if not finish[i]:
            # Check if process requirements can be satisfied
            if all(need[i][j] <= work[j] for j in range(m)):

                print(f"\nProcess P{i} can execute.")
                print("Need:", need[i])
                print("Available before:", work)

                # Release resources after process completion
                for j in range(m):
                    work[j] += allocation[i][j]

                finish[i] = True
                safe_sequence.append(i)

                print("Available after:", work)
                found = True

    if not found:
        break

# Display final result
print("\n" + "=" * 50)

if len(safe_sequence) == n:
    print("RESULT: SYSTEM IS IN A SAFE STATE")
    print("\nSafe Sequence:")
    print(" -> ".join(f"P{i}" for i in safe_sequence))
else:
    print("RESULT: SYSTEM IS IN AN UNSAFE STATE")
    unfinished = [f"P{i}" for i in range(n) if not finish[i]]
    print("Processes that could not finish:", unfinished)

print("=" * 50)
```

```python

# Step 6: Input Validation and Error Handling

print("=" * 50)
print(" DEADLOCK DETECTION AND AVOIDANCE SIMULATOR ")
print("          INPUT VALIDATION")
print("=" * 50)

# Input number of processes and resources
n = int(input("Enter number of processes: "))
m = int(input("Enter number of resource types: "))

# Validate process and resource count
if n <= 0 or m <= 0:
    raise ValueError("Number of processes and resources must be positive.")

# Input Allocation Matrix
allocation = []

print("\nEnter Allocation Matrix:")
for i in range(n):
    row = list(map(int, input(f"Allocation for P{i}: ").split()))

    if len(row) != m:
        raise ValueError(f"Enter exactly {m} values for P{i}.")

    if any(value < 0 for value in row):
        raise ValueError("Allocation values cannot be negative.")

    allocation.append(row)

# Input Maximum Matrix
maximum = []

print("\nEnter Maximum Matrix:")
for i in range(n):
    row = list(map(int, input(f"Maximum for P{i}: ").split()))

    if len(row) != m:
        raise ValueError(f"Enter exactly {m} values for P{i}.")

    if any(value < 0 for value in row):
        raise ValueError("Maximum values cannot be negative.")

    maximum.append(row)

# Check Allocation does not exceed Maximum
for i in range(n):
    for j in range(m):
        if allocation[i][j] > maximum[i][j]:
            raise ValueError(
                f"Error: Allocation exceeds Maximum for P{i}, resource {j}."
            )

# Input Available Resources
available = list(map(int, input(
    "\nEnter available resources separated by spaces: "
).split()))

if len(available) != m:
    raise ValueError(f"Enter exactly {m} available resource values.")

if any(value < 0 for value in available):
    raise ValueError("Available resources cannot be negative.")

# Calculate Need Matrix
need = []

for i in range(n):
    row = []
    for j in range(m):
        row.append(maximum[i][j] - allocation[i][j])
    need.append(row)

# Display validated data
print("\nInput validated successfully!")

print("\nNeed Matrix:")
for i, row in enumerate(need):
    print(f"P{i}: {row}")

print("\nAvailable Resources:", available)
print("\nAll inputs are valid.")



```

```python

# Deadlock Detection and Avoidance Simulator
# Using Banker's Algorithm

print("=" * 55)
print(" DEADLOCK DETECTION AND AVOIDANCE SIMULATOR ")
print("          BANKER'S ALGORITHM")
print("=" * 55)

# Step 1: Input number of processes and resources
try:
    n = int(input("\nEnter number of processes: "))
    m = int(input("Enter number of resource types: "))

    if n <= 0 or m <= 0:
        raise ValueError("Counts must be positive.")

    # Step 2: Input Allocation Matrix
    allocation = []
    print("\nEnter Allocation Matrix:")

    for i in range(n):
        row = list(map(int, input(f"P{i}: ").split()))

        if len(row) != m or any(x < 0 for x in row):
            raise ValueError("Invalid allocation row.")

        allocation.append(row)

    # Step 3: Input Maximum Matrix
    maximum = []
    print("\nEnter Maximum Matrix:")

    for i in range(n):
        row = list(map(int, input(f"P{i}: ").split()))

        if len(row) != m or any(x < 0 for x in row):
            raise ValueError("Invalid maximum row.")

        maximum.append(row)

    # Step 4: Validate Allocation and Maximum
    for i in range(n):
        for j in range(m):
            if allocation[i][j] > maximum[i][j]:
                raise ValueError(
                    f"Allocation exceeds Maximum for P{i}."
                )

    # Step 5: Input Available Resources
    available = list(map(int, input(
        "\nEnter available resources: "
    ).split()))

    if len(available) != m or any(x < 0 for x in available):
        raise ValueError("Invalid available resources.")

    # Step 6: Calculate Need Matrix
    need = []

    for i in range(n):
        row = []
        for j in range(m):
            row.append(maximum[i][j] - allocation[i][j])
        need.append(row)

    # Display matrices
    print("\nAllocation Matrix:")
    for row in allocation:
        print(row)

    print("\nMaximum Matrix:")
    for row in maximum:
        print(row)

    print("\nNeed Matrix:")
    for row in need:
        print(row)

    print("\nAvailable Resources:", available)

    # Step 7: Banker's Safety Algorithm
    work = available.copy()
    finish = [False] * n
    safe_sequence = []

    print("\nChecking system safety...")

    while len(safe_sequence) < n:
        found = False

        for i in range(n):
            if not finish[i] and all(
                need[i][j] <= work[j] for j in range(m)
            ):
                print(f"\nP{i} can complete.")
                print("Available before:", work)

                for j in range(m):
                    work[j] += allocation[i][j]

                finish[i] = True
                safe_sequence.append(i)

                print("Available after:", work)
                found = True

        if not found:
            break

    # Step 8: Final Result
    print("\n" + "=" * 55)

    if len(safe_sequence) == n:
        print("RESULT: SYSTEM IS IN A SAFE STATE")
        print("Safe Sequence:",
              " -> ".join(f"P{i}" for i in safe_sequence))
    else:
        print("RESULT: SYSTEM IS IN AN UNSAFE STATE")
        unfinished = [
            f"P{i}" for i in range(n) if not finish[i]
        ]
        print("Processes unable to complete:", unfinished)

    print("=" * 55)

except ValueError as error:
    print("\nInput Error:", error)
```

```python

# Step 9: Professional Output Formatting

def display_matrix(title, matrix, n, m):
    print("\n" + "=" * 45)
    print(title)
    print("=" * 45)

    header = "Process".ljust(12)
    for j in range(m):
        header += f"R{j}".ljust(8)

    print(header)
    print("-" * 45)

    for i in range(n):
        row_text = f"P{i}".ljust(12)

        for j in range(m):
            row_text += str(matrix[i][j]).ljust(8)

        print(row_text)

# Display matrices in table format
display_matrix("ALLOCATION MATRIX", allocation, n, m)
display_matrix("MAXIMUM MATRIX", maximum, n, m)
display_matrix("NEED MATRIX", need, n, m)

print("\n" + "=" * 45)
print("AVAILABLE RESOURCES")
print("=" * 45)

for j in range(m):
    print(f"Resource R{j}: {available[j]}")

print("=" * 45)
```
